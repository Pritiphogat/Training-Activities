```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# 1. Create a DataFrame with four columns of random numerical values
np.random.seed(42)
df = pd.DataFrame({
    "f1": np.random.randint(10, 101, 10),
    "f2": np.random.randint(10, 101, 10),
    "f3": np.random.randint(10, 101, 10),
    "f4": np.random.randint(10, 101, 10)
})
```

```python

# 2. Generate a scatter plot showing the relationship between f2 and f3
plt.scatter(df["f2"], df["f3"])
plt.xlabel("f2")
plt.ylabel("f3")
plt.title("Relationship between f2 and f3")
plt.grid(True, alpha=0.3)
plt.show()

# 3. Calculate the row-wise arithmetic mean and store it in f5
df["f5"] = df[["f1", "f2", "f3", "f4"]].mean(axis=1)

# 4. Export all five columns to an Excel file
df.to_excel("Pandas_Practical_Lab.xlsx", index=False)

print(df)
print("Excel file exported successfully.")
```
